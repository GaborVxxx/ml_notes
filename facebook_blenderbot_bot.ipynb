```python
from transformers import AutoTokenizer, AutoModelForSeq2SeqLM, pipeline

model_name = "facebook/blenderbot-400M-distill"

tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModelForSeq2SeqLM.from_pretrained(model_name)

conversation_history = ["You are a helpfull chat bot. That help the people to find the right answer."]  # Initialize with a starting message

def get_response(prompt):
    history_string = "\n".join(conversation_history)
    inputs = tokenizer.encode_plus(history_string, prompt, return_tensors="pt")  # Changed variable name
    output = model.generate(**inputs, max_length=128)  # Increased max_length
    response = tokenizer.decode(output[0], skip_special_tokens=True).strip()
    return response

prompt = "Do you have any idea for new digital product?"
response = get_response(prompt)
print(response)

conversation_history.append(prompt)
conversation_history.append(response)


```
